```python
# gradebook: rows=students, cols=Math,Sci,Eng
gradebook = np.array([
    [88, 75, 91],
    [92, 89, 85],
    [79, 91, 72],
])

# Global aggregations (all elements)
print(np.sum(gradebook))       # 762
print(np.mean(gradebook))      # 84.67

# axis=0: aggregate across rows → per-subject stats
print(np.mean(gradebook, axis=0)) # [86.33 85. 82.67]

# axis=1: aggregate across cols → per-student stats
print(np.mean(gradebook, axis=1)) # [84.67 88.67 80.67]
print(np.std(gradebook,  axis=1)) # std per student
```

```python
# Dot product (1D): sum of element-wise products
weights = np.array([0.4, 0.35, 0.25])  # exam weights
ali     = np.array([88, 75, 91])         # Ali's scores
final   = np.dot(weights, ali)           # weighted avg
print(f"Ali final: {final:.2f}")         # 83.45

# Matrix multiplication: gradebook × weights
gradebook = np.array([[88,75,91],[92,89,85],[79,91,72]])
finals    = gradebook @ weights          # or np.matmul()

# Inverse and determinant
A   = np.array([[2,1],[5,3]], dtype=float)
A_inv = np.linalg.inv(A)
print(np.linalg.det(A))

# Solve Ax = b (linear system)
b = np.array([1, 2])
x = np.linalg.solve(A, b)
print(x)
```

```python
gradebook = np.array([[88, 75, 91], [92, 89, 85], [79, 91, 72]])
weights   = np.array([0.4, 0.35, 0.25])

# --- Save ---
# savetxt writes a text file and returns None -> don't assign it to a variable
# comments="" -> header line is written as "Math,Science,English" instead of
# "# Math,Science,English", so pandas.read_csv gets clean column names later
np.savetxt("gradebook.csv", gradebook,
           delimiter=",", fmt="%d",
           header="Math,Science,English", comments="")

# save the ARRAY itself, not the return value of savetxt
np.save("gradebook.npy", gradebook)

# savez puts several arrays into one .npz file (needed for the np.load below)
np.savez("semester.npz", gradebook=gradebook, weights=weights)

# --- Load back ---
loaded    = np.load("gradebook.npy")
multi     = np.load("semester.npz")
# header is now plain text (no '#'), so it must be skipped explicitly
from_csv  = np.loadtxt("gradebook.csv", delimiter=",", skiprows=1)

print(loaded)
print(multi["weights"])
print(from_csv)
```
