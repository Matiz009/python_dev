```python
class Person:
    def __init__(self, initial_balance: float = 0.0):
        # Private attribute (name mangling)
        self.__balance = initial_balance

    def deposit(self, amount: float) -> None:
        """Deposits a positive amount into the account balance."""
        if amount <= 0:
            raise ValueError("Deposit amount must be positive.")
        self.__balance += amount

    def withdraw(self, amount: float) -> None:
        """Withdraws an amount if sufficient funds are available."""
        if amount <= 0:
            raise ValueError("Withdrawal amount must be positive.")
        if amount > self.__balance:
            raise ValueError("Insufficient funds.")
        self.__balance -= amount

    @property
    def balance(self) -> float:
        """Getter property to read the balance safely."""
        return self.__balance


# Usage
ali = Person(50000)
ali.deposit(2000)

print(f"Ali's current balance: {ali.balance}")  # Output: 52000

ali.withdraw(23000)
```

```python
ali.balance
```

```python
class Vehicle:
    def __init__(self, brand):
        self.brand = brand
    def start(self):                 # shared behavior
        return f'{self.brand} starting'
```

```python
Audi = Vehicle('Audi')
```

```python
Audi.start()
```

```python
Audi.brand
```
