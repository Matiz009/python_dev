# Faqs dictionary

```python
faq = {
    "what time does the lab open": "It opens at 9 AM, Monday to Friday.",
    "how do i submit my assignment": "Upload it to the class portal before 11:59 PM.",
    "when is the next exam": "The next exam is on the 20th of this month.",
    "who is my instructor": "Your instructor is Sir Mati.",
    "where is the classroom": "Room 204, second floor, Block B.",
}
```

## Creating a function that will match and calculate the best possible score

```python
from difflib import SequenceMatcher

def find_best_match(user_question, faq):
    user_question = user_question.lower().strip()
    best_key, best_score = None, 0

    for question in faq:
        score = SequenceMatcher(None, user_question, question).ratio()
        if score > best_score:
            best_key, best_score = question, score

    return best_key, best_score
```

```python
THRESHOLD = 0.6

def get_answer(user_question, faq):
    key, score = find_best_match(
        user_question, faq)

    if score >= THRESHOLD:
        return faq[key]
    else:
        return ("I don't know. "
                "Please ask your instructor.")
```

```python
print("Hi! Ask me a question (type 'quit' to exit).")

while True:
    user_input = input("You: ")
    if user_input.lower() == "quit":
        print("Bot: Goodbye!")
        break

    answer = get_answer(user_input, faq)
    print("Bot:", answer)
```
