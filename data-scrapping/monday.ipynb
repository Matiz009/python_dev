```python
import time

base_url = "http://quotes.toscrape.com/page/{}/"
all_quotes = []

for page in range(1, 11):
    r = requests.get(base_url.format(page))
    soup = BeautifulSoup(r.text, 
"html.parser"
)
    quotes = soup.find_all(
"div", class_="quote"
)
    if not quotes:
        break
    all_quotes.extend(quotes)
    time.sleep(1)

print(len(all_quotes))
```
