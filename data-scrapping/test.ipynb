```python
import requests
from bs4 import BeautifulSoup
import csv
```

```python

url = "http://quotes.toscrape.com"
soup = BeautifulSoup(requests.get(url).text, "html.parser")

# one-liner using class name
quote = soup.select_one(".quote .text").get_text(strip=True)
author = soup.select_one(".quote .author").get_text(strip=True)

print(quote, "-", author)
```

```python
# one-liner using find()
quote = soup.find("span", class_="text").get_text(strip=True)
author = soup.find("small", class_="author").get_text(strip=True)

print(quote, "-", author)
```

```python
# using find_all (class-based)
tags = [t.get_text(strip=True) for t in soup.find_all("div", class_="quote")]
print(tags)
```
