```python
# Series — single labeled column
scores = pd.Series([88, 92, 79, 95],index=["Ali","Sara","Bilal","Hina"], name="Math_Score")
```

```python
# DataFrame — full student record
df = pd.DataFrame({
    "student":       ["Ali","Sara","Bilal","Hina"],
    "Math_Score":    [88, 92, 79, 95],
    "Science_Score": [75, 89, 91, 84],
    "Grade":         [9, 10, 9, 11]
})
```

```python
print(df)
```

```python
print(df['Math_Score'].median())
```
